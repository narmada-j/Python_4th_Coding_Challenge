```python
# 10 - Function Arguments, Local and Global Namespace, Introduction to OOP
#-------------------------------------------------------------------------
#  Bank Account Manager
#-----------------------
#3. global variables, such as a global transaction counter.
transaction_counter = 0
#1. Create a class BankAccount with account_holder and balance attributes.
class BankAccount:
    def __init__(self, account_holder, balance=0):
        self.account_holder = account_holder
        self.balance = balance
#2. Implement methods for deposit(), withdraw(), and display_balance().
    def deposit(self, amount):
        global transaction_counter
        if amount > 0:
            self.balance += amount
            transaction_counter += 1
            print(f"Deposited {amount}. New balance: {self.balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
      #3. Use both local and global variables appropriately, such as a global transaction counter.
        global transaction_counter
        if amount <= 0:
            print("Withdrawal amount must be positive.")
        elif amount > self.balance: #5. Prevent withdrawals that would result in a negative balance.
            print("Insufficient funds! Withdrawal denied.")
        else:
            self.balance -= amount
            transaction_counter += 1
            print(f"Withdrew {amount}. New balance: {self.balance}")

    def display_balance(self):
        print(f"Account Holder: {self.account_holder}, Balance: {self.balance}")

#4. Demonstrate calling methods using both positional and keyword arguments.
name = input("Enter account holder name: ")
initial_balance = float(input("Enter initial balance: "))

# Create account using user input
account = BankAccount(account_holder=name, balance=initial_balance)

# Perform transactions interactively
while True:
    print("\nChoose an option:")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Display Balance")
    print("4. Exit")

    choice = input("Enter choice (1-4): ")

    if choice == "1":
        amount = float(input("Enter deposit amount: "))
        account.deposit(amount)
    elif choice == "2":
        amount = float(input("Enter withdrawal amount: "))
        account.withdraw(amount)
    elif choice == "3":
        account.display_balance()
    elif choice == "4":
        print("Exiting... Thank you!")
        break
    else:
        print("Invalid choice. Please try again.")

print(f"\nTotal transactions made: {transaction_counter}")
```
