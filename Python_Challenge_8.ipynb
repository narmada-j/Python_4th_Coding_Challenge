```python
# 8 -  range, for loop, enumerate, list comprehension
#----------------------------------------------------
# Employee Hours Report
#-----------------------
import random
#1. Create a list of 10 random exam scores ranging from 0 to 100 using list comprehension.
print("Random Exam Score")
exam_scores = [random.randint(0, 100) for _ in range(10)]
print(exam_scores)
#2. Display each score with its index using the enumerate function.
for index,score in enumerate(exam_scores):
  print(index,score)
#3. Find all scores greater than 75 and display them.
print("\nValues Greater than 75")
for i in exam_scores:
  if i > 75:
    print(i)
#4. Calculate and display the average score.
avg_score = sum(exam_scores) / len(exam_scores)
print("\nAverage Score:", avg_score)
```
