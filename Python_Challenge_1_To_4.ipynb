```python
# 1 A- Student Profile Creator
#-----------------------------

# 1. Create variables to store student details
student_name = "Arjun"
student_age = 15
fav_subjects = ["Mathematics", "Science", "English"]

# 2. Display the student’s information in a user-friendly format
print(" Student Profile ")
print("-------------------")
print(f"Name: {student_name}")
print(f"Age: {student_age}")
print("Favourite Subjects:")
for subject in fav_subjects:
    print(f" * {subject}")

# 3. Update the list of subjects (adding a new interest)
fav_subjects.append("Computer Science")

print("\n Updated Favourite Subjects:")
print(" ---------------------------")
for subj in fav_subjects:
    print(f" * {subj}")
print("\n")
```

```python
# 1 B- Student Profile Creator
#-----------------------------

# 1. Get student details from user
student_name = input("Enter the student's name: ")
student_age = int(input("Enter the student's age: "))

# 2. Get number of subjects
num_subjects = int(input("\nEnter the number of favourite subjects: "))

# Create an empty list
fav_subjects = []

# Loop to collect subject names
for i in range(num_subjects):
    subject = input(f"Enter subject {i+1}: ")
    fav_subjects.append(subject)

# 3. Display the student’s information in a user-friendly format
print("\n Student Profile ")
print("-------------------")
print(f"Name: {student_name}")
print(f"Age: {student_age}")
print("Favourite Subjects:")
for subject in fav_subjects:
    print(f" * {subject}")

# 4. Update the list of subjects (adding a new interest)
new_subject = input("\nEnter a new subject to add: ")
fav_subjects.append(new_subject)

print("\n Updated Favourite Subjects:")
print(" ---------------------------")
for subj in fav_subjects:
    print(f" * {subj}")
print("\n")
```

```python
# 2 -  Feedback Analyzer
#-----------------------------
#input: Data set analysis was insightful, Python slicing improved clarity, and string methods made data cleaning faster and more efficient value

# 1. Ask the user to input a sentence (representing feedback).
feedback = input("Enter Your Feedback: ")
print("----------------")
print("|   Feedback   |")
print("----------------")
print(feedback)

# 2. Count how many words are in the sentence.
wordList = feedback.split(" ")    # splitted values are storing in list
word_count=len(wordList)
print(f"\nNo Of Words:{word_count}")

# 3. Check how many times the word “data” appears in the sentence, regardless of case.
times_count = feedback.lower().count("data")
print(f"\nNo Of Time Data Appears: {times_count}")

# 4. Extract and print the first 5 and last 5 characters of the feedback separately.
first5 = feedback[:5]
last5 = feedback[-5:]
print(f"\nFirst 5 Characters:{first5}")
print(f"\nLast 5 Characters:{last5}")

# 5. Reverse the sentence and print it.
reversed = feedback[::-1]
print(f"\nReversed Feedback: {reversed}\n")
```

```python
# 3 -  Tuple, Immutable vs Mutable, List Creation
#------------------------------------------------

# 1. Create a tuple with fixed product categories like Electronics, Stationery, and Clothing.
category = ('Electronics', 'Stationery', 'Clothing')
print(f"Tuple : {category}")

# 2. Create a list of products that can be updated dynamically.
product = list(category)
print(f"\nList  : {product}")
# 3. Ask the user to enter a new product and add it to the list.
new = input("Enter category to Add      : ")
category = category + (new,)
print(f"\nTuple After New Item Added : {category}")
# 4. Ask the user to input the index of a product to remove it from the list.
product1 = list(category)
print("\n   Tuple Values\n----------------")
for index, value in enumerate(product1):
    print(f"Index {index}: {value}")
#print("Enter index Number to remove values from Tuple : ")
ind = int(input("Enter index Number to remove values from Tuple :"))
product1.pop(ind)
category = tuple(product1)
print(f"Tuple After Removed : {category}")
# 5. Display the final list of products and the product categories.
print(f"\nTuple : {category}")
print(f"List  : {product}")

```
