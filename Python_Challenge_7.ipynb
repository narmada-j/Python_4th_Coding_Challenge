```python
# 7 - While Loop, else, break, continue
#----------------------------------------
# Shopping Cart System
#----------------------
#Continuously accept item prices using a while loop.
item = 1
items = []
subtot= 0
while True:
   amt = int(input(f"Enter Item {item} Amount: "))
   #Stop taking input when the user enters 0 (sentinel value).
   if amt == 0:
    break
   # Ensure no negative prices are allowed (ask again without counting as an item).
   elif amt < 0:
     print("Please add correct amount....!")
     continue
   else:
    # Keep track of the total number of items and the subtotal amount.
    items.append(amt)
    subtot +=amt
    item+=1
    continue

# Apply discounts based on the subtotal
disc = 0
if subtot < 200:
  disc = 0
elif subtot >= 200 and subtot <= 500:
  disc = 5
elif subtot >= 500 and subtot <= 1000:
  disc = 10
elif subtot >= 1000:
  disc = 20
# Calculate discount amount,final amount
dis_amt = 0
final_amt = 0
dis_amt = subtot * disc / 100
final_amt = subtot - dis_amt
# Display a clear bill summary including subtotal, discount amount, and final total.
print("\nBill Summary\n----------------------")
print("Total Items     : ",len(items))
print("SubTotal        : ",subtot)
print("Discount(%)     : ",disc)
print("Discount Amount : ",dis_amt)
print("Final Amount    : ",final_amt)
```
