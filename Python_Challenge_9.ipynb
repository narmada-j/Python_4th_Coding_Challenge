```python
# 9 -  Built-in Functions and User-Defined Functions
#----------------------------------------------------
#  Simple Calculator App
#-----------------------
#1. Create functions for addition, subtraction, multiplication, and division.
def add(a, b):
    return a + b
def sub(a, b):
    return abs(a - b)
def mult(a, b):
    return a * b
def div(a, b):
    return round(a / b,2)
#2. Ask the user to input two numbers and the operation they want to perform.
a = int(input("Enter value 1: "))
b = int(input("Enter value 2: "))
#3. Call the appropriate function and display the result.
#4. Use built-in functions like abs(), round(), and pow() within your solution.
#5. Handle division by zero gracefully and inform the user.
choice = input("\n1.Add\n2.Subract\n3.Multiplication\n4.Divide\nChoose any one option: ")
match choice:
  case "1":
    print("Sum of value is: ",add(a,b))
  case "2":
    print("Subraction of value is: ",sub(a,b))
  case "3":
    print("Multiplication of value is: ",mult(a,b))
  case "4":
    if b > 0 :
      print("Division value is: ",div(a,b))
    else:
      print("Zero not be divided...!")
  case _:
    print("Invalid Option...!")
```
