```python
# 5 -  Dictionary, Set, Frozenset
#---------------------------------
#Student Grades & Hobbies Tracker
#--------------------------------
# 1. Create a dictionary with student names as keys and their grades as values.
student = {"Aarthi":"A","Bharath":"B","Chitra":"A","Eswari":"C"}
print(student)
#2. Allow the user to add or update grades for at least 3 students.
for i in range(3):
  choice = input("Enter 1.Add 2.Update (1-2): ")
  match choice:
      case "1":
          #adding new student
          name = input("Enter Student Name : ")
          grade = input("Enter Grade Of Student : ")
          student[name] = grade
      case "2":
          #update grade
          print("\nKey   :   Values\n--------------------")
          for key,val in student.items():
            print(f"{key} : {val}")
          upd_key=input("Type Key Name to update Grade: ")
          upd_val = input("Enter Grade to update: ")
          if upd_key in student:
            student[upd_key] = upd_val
          else:
            print("Key not Found...!")
            print(student)
      case _:
          print("Invalid choice")

  print("Updated: ",student)
```

```python
#3. Create a set of hobbies and allow the user to input at least 3 hobbies.
set1 = {"Music","Running","Learning"}
print("Intial set: ",set1)
for i in range(3):
  hob = input("Enter Hobbies to add: ")
  set1.add(hob)
print("Updated set: ",set1)
#4. Convert the set of hobbies into a frozenset and display it.
frozen_hobbies = frozenset(set1)
print("Frozenset:", frozen_hobbies)
```
