```python
# 6 - Operators, Conditional Statements
#----------------------------------------
# BMI Calculator
#----------------
#1. Ask the user to enter their weight (in kilograms) and height (in meters).
weight = float(input("\nEnter Your Weight(kg): "))
height = float(input("\nEnter Your Height(Meters): "))
#2. Calculate their BMI using the formula BMI = weight / (height × height).
bmi = weight / (height ** 2)
#3. Classify the BMI as underweight, normal, overweight, or obese using conditional statements.
#4. Display the result clearly.
if bmi < 18.5:
  print("Underweight")
elif bmi > 18.5 and bmi < 24.9:
  print("Normal Weight")
elif bmi > 24.9 and bmi < 29.9:
  print("Over Weight")
elif  bmi > 30.0:
  print("Obese")
else:
  print("Invalid")
```
